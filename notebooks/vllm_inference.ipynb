```python
from vllm import LLM, SamplingParams
from PIL import Image
import GPUtil
import time
```

```python
from vllm import LLM
import inspect

print(inspect.signature(LLM))
```

```python
def sample_gpu():
    gpu = GPUtil.getGPUs()[0]
    return {
        "util": gpu.load * 100,
        "mem": gpu.memoryUtil * 100,
        "mem_used": gpu.memoryUsed
    }
```

```python
llm = LLM(
    model="llava-hf/llava-1.5-7b-hf",
    trust_remote_code=True,
    max_model_len=4096,
    gpu_memory_utilization=0.9,
)

sampling_params = SamplingParams(
    temperature=0.7,
    top_p=0.9,
    max_tokens=50
)
```

```python
prompts = [
    "USER: <image>\nWhat do you see in this image?\nASSISTANT:",
    "USER: <image>\nDescribe this image in detail.\nASSISTANT:",
    "USER: <image>\nWhat colors are prominent?\nASSISTANT:"
]

image = Image.open("test_image.jpg")
```

```python
batch_inputs = [
    {
        "prompt": prompts[i % len(prompts)],
        "multi_modal_data": {"image": image}
    }
    for i in range(1000)
]
```

```python
samples = []
start = time.time()
samples.append(("before", time.time() - start, sample_gpu()))

outputs = llm.generate(batch_inputs, sampling_params=sampling_params)

end_time = time.time() - start
samples.append(("after", end_time, sample_gpu()))
```

```python
print(f"Total Time: {end_time:.2f}s")
print(f"Time per request: {end_time/len(batch_inputs)*1000:.2f}ms")

print("GPU Stats:")
print(f"Before - Util: {samples[0][2]['util']:.1f}%, Mem: {samples[0][2]['mem']:.1f}%, Used: {samples[0][2]['mem_used']}MB")
print(f"After  - Util: {samples[1][2]['util']:.1f}%, Mem: {samples[1][2]['mem']:.1f}%, Used: {samples[1][2]['mem_used']}MB")
```
